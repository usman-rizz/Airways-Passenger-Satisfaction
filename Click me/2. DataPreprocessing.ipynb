### <b>Converting Rating Columns to int64</b>

All rating columns (service quality, convenience ratings, etc.) have values in the range 0–5, so we convert them to int64 for memory efficiency and consistency.

```python
# List of rating columns to convert to int64

rating_columns = [
    'Inflight wifi service',
    'Departure/Arrival time convenient',
    'Ease of Online booking',
    'Gate location',
    'Food and drink',
    'Online boarding',
    'Seat comfort',
    'Inflight entertainment',
    'On-board service',
    'Leg room service',
    'Baggage handling',
    'Checkin service',
    'Inflight service',
    'Cleanliness'
]

# Convert each rating column to int64
for col in rating_columns:
    cleaned_data[col] = cleaned_data[col].astype('Int64')
    print(f"✓ Converted '{col}' to int64")

print(f"\nTotal columns converted: {len(rating_columns)}")

print()

print("\nMemory usage after converting rating columns to int64:")
cleaned_data.info()
```

### <b>Convert multiple columns to category using a loop</b>

```python
cat_cols = ['Customer Type', 'Gender', 'Type of Travel', 'Class']

for col in cat_cols:
    if col in cleaned_data.columns:
        cleaned_data[col] = cleaned_data[col].astype('category')
        print(f"✓ Converted '{col}' to category")
    else:
        print(f"⚠️ Column '{col}' not found in dataframe")


print('\nDtypes for converted columns:')
print(cleaned_data[cat_cols].dtypes)
print("\n")
cleaned_data.info()
```

# **Imputation Methods:**

#### 1. **Mean/Median Imputation:**
- If there are missing values in the Age column, you can replace them with the column’s mean or median value.
- This method is simple but ignores other influencing factors.

#### 2. **Group-wise Imputation:**
- You can impute Age based on groups such as **Pclass** or **Sex**.
- For example, calculate the median age for each Pclass and Sex group and replace missing values with that group’s median.
- This method is more accurate as it considers other factors.

#### 3. **Machine Learning Models:**
- You can use machine learning models like **KNNImputer** or **RandomForestRegressor** to predict missing ages based on other features.
- This method is the most accurate but requires more computational resources and time.

### **Imputation of All Rating Columns with groupby() wrt Travel Type Column** 

```python
# Show the most frequent value for each rating column per 'Type of Travel'
most_frequent_per_group = cleaned_data_2.groupby('Type of Travel')[rating_columns] \
                                        .agg(lambda x: x.mode()[0])
print("Most frequent values per group:\n")
most_frequent_per_group
```

```python
from sklearn.impute import SimpleImputer

# Function to fill missing values in all rating columns for each group
def fill_groupwise_most_frequent(group, columns):
    imputer = SimpleImputer(strategy='most_frequent')
    group[columns] = imputer.fit_transform(group[columns]).astype(int)
    return group

# Reset index to avoid groupby ambiguity
cleaned_data_2 = cleaned_data_2.reset_index(drop=True)

# Apply the function to each 'Type of Travel' group
cleaned_data_2 = cleaned_data_2.groupby('Type of Travel', group_keys=False) \
                               .apply(fill_groupwise_most_frequent, columns=rating_columns)
```
